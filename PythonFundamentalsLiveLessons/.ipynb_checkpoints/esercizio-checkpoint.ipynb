minimo, massimo, media, somma di liste. fare funzioni, chiamarle e stampare risultato. con while e for

```python
def elevato(lista1):
    for i in range(0, len(lista1)):
        lista1[i] = lista1[i] ** 2
    return lista1
elevato([3, 5, 6, 4])
```

```python
def numero_minimo(lista3):
    minimo = lista3[0]
    for numero in lista3:
        if numero < minimo:
            minimo = numero
    return minimo
```

```python
numero_minimo([3, 6, 90, 69, 47])
```

```python
def trova_minimo(lista5):
    minimo = lista5[0]
    i = 1
    while i < len(lista5):
        if lista5[i] < minimo:
            minimo = lista5[i]
        i += 1
    return minimo
```

```python
trova_minimo([3, 6, 9, -5, 11])
```

```python
def numero_massimo(lista4):
    massimo = lista4[0]
    for numero in lista4:
        if numero > massimo:
            massimo = numero
    return massimo
```

```python
numero_massimo([4, 6, 7, -5, 11])
```

```python
trova_massimo([4, 6, 9, 30, 69, 3, 1, -30])
```
