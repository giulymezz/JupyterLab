```python
def maximum(value1, value2, value3):
    max_value = value1
    if value2 > max_value:
        max_value = value2
    if value3 > max_value:
        max_value = value3
    return max_value
```

```python
maximum(22, 4, 17)
```

```python
max(22, 594, 963, 535, 93.654)
```

```python
import random

# Inizializza i contatori
face1 = face2 = face3 = face4 = face5 = face6 = 0

for _ in range(6_000_000):
    face = random.randrange(1, 7)
    if face == 1:
        face1 += 1
    elif face == 2:
        face2 += 1
    elif face == 3:
        face3 += 1
    elif face == 4:
        face4 += 1
    elif face == 5:
        face5 += 1
    else:
        face6 += 1

# Stampa dei risultati
print(f"{'Face':<5}{'Frequency':>13}")
print(f"{'1':<5}{face1:>13}")
print(f"{'2':<5}{face2:>13}")
print(f"{'3':<5}{face3:>13}")
print(f"{'4':<5}{face4:>13}") 
print(f"{'5':<5}{face5:>13}")
print(f"{'6':<5}{face6:>13}")
```

```python
import math

# Calcolare la radice quadrata di 900
print(math.sqrt(900))  # Output: 30.0

# Calcolare il valore assoluto di un numero negativo (float)
print(math.fabs(-15.5))  # Output: 15.5

# Usare la costante pi greco
print(math.pi)  # Output: 3.141592653589793

# Usare la costante e (numero di Eulero)
print(math.e)  # Output: 2.718281828459045
```

```python
# Definizione della funzione
def rectangle_area(length=2, width=3):
    """Restituisce l'area di un rettangolo."""
    return length * width
```

```python
# Nessun argomento:
print(rectangle_area())  # Output: 6 (usa 2 * 3)
```

```python
# Un solo argomento:
print(rectangle_area(10))  # Output: 30 (usa 10 * 3)
```

```python
# Due argomenti:
print(rectangle_area(4, 5))  # Output: 20 (usa 4 * 5)
```

```python
def rectangle_area(length, width):
    return length * width
rectangle_area(10, 3)  # Restituisce 30
```

```python
rectangle_area(width=3, length=10)  # Restituisce 30
```

```python
x = 7  # variabile globale

def stampa_x():
    print("X (dallo scope globale):", x)

stampa_x()  # Output: 7
```

```python
x = 7

def cambia_x():
    x = 3.5  # crea una nuova variabile locale
    print("X dentro la funzione:", x)

cambia_x()       # Output: 3.5
print("X globale:", x)  # Output: 7 (non è stata modificata)
```

```python
x = 7

def modifica_globale():
    global x
    x = "hello"
    print("X modificata nella funzione:", x)

modifica_globale()       # Output: hello
print("X globale ora è:", x)  # Output: hello
```

```python
def cube(number):
    print("ID dentro la funzione (prima):", id(number))
    number = number ** 3
    print("ID dentro la funzione (dopo):", id(number))
    return number

x = 7
print("ID fuori dalla funzione:", id(x))
result = cube(x)
print("ID di x dopo la funzione:", id(x))
print("Valore di x:", x)
print("Valore di result:", result)
```

```python
def append_item(lst):
    print("ID dentro la funzione:", id(lst))
    lst.append(4)

my_list = [1, 2, 3]
print("ID fuori dalla funzione:", id(my_list))
append_item(my_list)
print("Lista dopo la funzione:", my_list)
```
