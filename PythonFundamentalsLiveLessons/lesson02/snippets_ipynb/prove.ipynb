```python
total = 0
count = 0

grade = int(input("Inserisci un voto (-1 per terminare): "))

while grade != -1:
    total += grade
    count += 1
    grade = int(input("Inserisci un voto (-1 per terminare): "))

if count != 0:
    average = total / count
    print(f"La media è: {average:.2f}")
else:
    print("Nessun voto inserito.")
```

```python
from decimal import Decimal

principle = Decimal("1000")
rate = Decimal("0.05")

for year in range(1, 11):
    amount = principle * (1 + rate) ** year
    print(f"Anno {year}: €{amount:.2f}")
```

```python
from decimal import Decimal

principle = Decimal("1000")
rate = Decimal("0.05")

for year in range(1, 11):
    amount = principle * (1 + rate) ** year
    print(f"Anno {year:>2}: €{amount:>10.2f}")
```
