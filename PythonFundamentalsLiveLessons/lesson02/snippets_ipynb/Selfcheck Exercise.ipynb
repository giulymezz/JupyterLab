#### Ex 2.9

##### For the values 47, 95, 88, 73, 88 and 84 calculate the minimum, maximum and range.

`````python
numeri = [47, 95, 88, 73, 88, 84]

minimo = min(numeri)
massimo = max(numeri)
range = max(numeri) - min(numeri)

print(f"Il numero minimo è {minimo}.")
print(f"Il numero massimo è {massimo}.")
print(f"Il range dei numeri è {range}.")
`````
