#### Ex 3.14

##### Assume that the tax on a restaurant bill is 6.25% and that the bill amount is $37.45. Use type Decimal to calculate the bill total then print the result with two digits to the right of the decimal point. 

````python
from decimal import Decimal

bill_amount = Decimal("37.45")
tax_rate = Decimal("0.0625")

# Calcolo della tassa
tax_amount = bill_amount * tax_rate

# Calcolo del totale
total = bill_amount + tax_amount

# Arrotondamento a due cifre decimali (stile finanziario)
total = total.quantize(Decimal("0.01"))

# Stampa del risultato
print(f"Total bill (with tax): ${total:.2f}")
````
